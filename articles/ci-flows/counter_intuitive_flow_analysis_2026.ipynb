## Case study: 2026's most widely counter-intuitive moments

We scan every Market Time Unit (MTU) across 2026, count how many borders are flagged `Counter-intuitive`, then take the timestamps with the highest counts.

```python
end = min(datetime.now(UTC), datetime(2027, 1, 1, tzinfo=UTC))

# We'll use VoltPath APIs to fetch flows data
flows_2026 = fetch_flows_range("2026-01-01", end.strftime("%Y-%m-%d"))

counts_2026 = hourly_border_counts(flows_2026, "Counter-intuitive")
total_borders = flows_2026["border"].nunique()

max_borders = counts_2026["n_borders"].max()
tied = counts_2026[counts_2026["n_borders"] == max_borders].sort_values("datetime")
print(
    (
        f"Peak: {max_borders} of {total_borders} borders simultaneously counter-intuitive, "
        f"{len(tied)} timestamp(s) tied at that max:"
    )
)
tied
```
